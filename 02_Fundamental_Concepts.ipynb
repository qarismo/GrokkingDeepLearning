## Ringkasan Chapter 2

Chapter 2 menjelaskan bagaimana pembelajaran mesin dipahami melalui beberapa konsep dasar. Deep learning ditempatkan sebagai bagian dari machine learning, sedangkan machine learning berada dalam bidang artificial intelligence. Neural network digunakan untuk mempelajari pola dan membantu menyelesaikan tugas praktis, tanpa mengharuskan sistem memiliki kecerdasan umum seperti manusia.

Pembahasan kemudian membedakan supervised dan unsupervised learning. Pada supervised learning, data masukan dipasangkan dengan keluaran target, sehingga model belajar memprediksi sesuatu yang diketahui pada contoh pelatihan. Pada unsupervised learning, tidak tersedia target seperti itu; buku menjelaskan pendekatan ini melalui pengelompokan data dan interpretasi struktur kelompok yang ditemukan.

Pengelompokan kedua membedakan parametric dan nonparametric learning. Buku mengibaratkan parameter sebagai tombol pengaturan yang menyimpan pola yang dipelajari. Model parametric mempunyai jumlah parameter yang ditetapkan untuk suatu bentuk model. Dalam penjelasan nonparametric, representasi yang diperlukan dapat mengikuti data. Analogi mencoba pengaturan dan menghitung frekuensi digunakan untuk membangun intuisi, tetapi bukan definisi yang mencakup seluruh algoritma.

Supervised parametric learning dijelaskan melalui tiga tahap: membuat prediksi, membandingkan prediksi dengan target, dan menyesuaikan parameter. Buku juga menunjukkan pengelompokan dengan parameter serta contoh penghitungan frekuensi pada lampu lalu lintas. Bab ini menjadi fondasi konseptual untuk implementasi neural network pada Chapter 3.

**Rujukan:** Trask (2019), hlm. 9-19.

## 4. What is deep learning?

Trask menempatkan deep learning sebagai bagian dari metode machine learning. Machine learning sendiri berada dalam bidang artificial intelligence. Hubungan ruang lingkupnya dapat ditulis:

$$
\mathrm{Deep\ Learning}\subset\mathrm{Machine\ Learning}\subset\mathrm{Artificial\ Intelligence}
$$

| Istilah | Penjelasan pengantar |
| --- | --- |
| Artificial intelligence | Bidang yang mengembangkan sistem untuk menjalankan tugas yang berkaitan dengan kecerdasan |
| Machine learning | Pendekatan yang memungkinkan sistem mempelajari pola dari data atau pengalaman |
| Deep learning | Kelompok pendekatan ML yang terutama menggunakan artificial neural network |

Implikasinya, tidak semua AI menggunakan machine learning, dan tidak semua machine learning merupakan deep learning. Dalam konteks buku, neural network menjadi alat utama yang akan dipelajari untuk mengolah gambar, teks, suara, dan jenis informasi lain.

**Pendalaman:** Nama artificial neural network menunjukkan inspirasi yang longgar dari sistem saraf, bukan salinan lengkap proses otak manusia. Kemampuan menyelesaikan tugas tertentu juga tidak otomatis menunjukkan kesadaran atau kecerdasan umum. Istilah deep biasanya berkaitan dengan pengolahan melalui beberapa lapisan; jaringan saraf sederhana tidak otomatis menjadi model deep hanya karena disebut neural network.

**Rujukan:** Trask (2019), hlm. 10. Penjelasan tentang istilah deep merupakan pendalaman pendamping.

## 5. What is machine learning?

Bab ini menjelaskan machine learning sebagai proses ketika mesin mengenali pola dan memanfaatkannya untuk menjalankan suatu tugas. Alih-alih menuliskan setiap aturan keluaran satu per satu, pengembang menyediakan mekanisme yang dapat memperoleh pola dari contoh atau data.

**Klarifikasi:** Belajar tanpa pemrograman aturan secara eksplisit bukan berarti komputer bekerja tanpa program. Manusia tetap menentukan cara merepresentasikan data, bentuk model, prosedur belajar, dan tujuan yang ingin dicapai. Yang dipelajari adalah pola untuk menyelesaikan tugas, bukan seluruh proses kerja komputer tanpa campur tangan manusia.

Salah satu contoh yang disebut buku adalah mempelajari hubungan antara suatu angka dan angka yang bernilai dua kali lipat. Contoh tersebut sederhana, tetapi memperlihatkan gagasan umum: informasi yang tersedia menjadi dasar untuk memperkirakan informasi yang ingin diketahui.

Pada masalah lain, hubungan masukan dan keluaran tidak sesederhana itu. Karena itu, model memerlukan contoh yang relevan, dan keberhasilan pada contoh yang telah dilihat belum dengan sendirinya menjamin keberhasilan pada data baru.

**Rujukan:** Trask (2019), hlm. 11. Klarifikasi tentang program dan data baru merupakan pendalaman tambahan.

## 6. Supervised machine learning

Supervised learning menggunakan contoh yang memiliki **masukan dan target yang berpasangan**. Model belajar hubungan keduanya, lalu digunakan untuk memperkirakan target bagi masukan lain.

Notasi pendamping untuk pasangan contoh adalah:

$$
\mathcal{D}=\{(x_i,y_i)\}_{i=1}^{n},\qquad \hat{y}_i=f_\theta(x_i)
$$

- $x_i$ adalah informasi masukan pada contoh ke-$i$.
- $y_i$ adalah target atau hasil yang diketahui untuk contoh tersebut.
- $\hat{y}_i$ adalah prediksi model.
- $\theta$ menyatakan parameter model pada contoh model parametric.
- $n$ menyatakan banyaknya contoh pelatihan.

Buku menggunakan ilustrasi hubungan data harga saham pada hari Senin dan hari Selasa. Inti ilustrasinya adalah struktur masukan-target, bukan bukti bahwa harga saham dapat diprediksi secara akurat atau saran untuk mengambil keputusan keuangan.

Walaupun buku mengibaratkan supervised learning sebagai transformasi satu dataset menjadi dataset lain, observasi masukan dan target harus cocok. Target dari satu objek tidak dapat dipasangkan sembarang dengan masukan dari objek lain.

**Pendalaman:** Pada tahap penggunaan model, target sering belum diketahui; itulah alasan prediksi diperlukan. Target tersedia pada contoh pelatihan untuk memberikan sinyal belajar dan pada data evaluasi untuk menilai hasil.

**Rujukan:** Trask (2019), hlm. 11-12. Notasi dan penjelasan pasangan observasi merupakan pendalaman tambahan.

## 7. Contoh supervised learning dan jenis target

| Informasi masukan | Target yang ingin dipelajari | Contoh bentuk tugas |
| --- | --- | --- |
| Piksel gambar | Ada atau tidak ada kucing | Klasifikasi |
| Kata-kata seseorang | Kategori emosi | Klasifikasi |
| Angka masukan | Angka dua kali lipat | Regresi |
| Rekaman suara | Transkrip kata | Prediksi urutan |

Contoh masukan dan target di atas diringkas dari ilustrasi buku. Kolom bentuk tugas merupakan penjelasan pendamping.

**Klasifikasi** memprediksi kategori, sedangkan **regresi** memprediksi besaran numerik. Supervised learning juga dapat menghasilkan keluaran yang lebih kompleks, seperti urutan kata. Karena itu, supervised learning tidak sebaiknya disamakan hanya dengan klasifikasi.

Misalnya, untuk menentukan ada atau tidaknya kucing, contoh pelatihan memerlukan informasi tentang status gambar. Untuk transkripsi suara, contoh pelatihan dapat memasangkan rekaman dengan teks yang sesuai. Pada kedua kasus, target menyediakan pola yang ingin dipelajari.

**Rujukan:** Trask (2019), hlm. 11-12. Pengelompokan jenis target merupakan pendalaman tambahan.

## 8. Unsupervised machine learning

Unsupervised learning mempelajari struktur dari data tanpa target berlabel seperti pada supervised learning. Pada penjelasan buku, struktur tersebut diperlihatkan melalui **clustering**, yaitu pengelompokan observasi berdasarkan pola kemiripan.

Algoritma dapat menghasilkan nomor kelompok bagi setiap observasi. Nomor itu merupakan identitas kelompok, bukan jawaban target yang sebelumnya disediakan untuk ditiru. Nama atau makna kelompok kemudian perlu ditafsirkan oleh manusia.

**Klarifikasi:** Buku menggunakan clustering sebagai gambaran utama unsupervised learning. Untuk pengertian yang lebih luas, unsupervised learning juga mencakup dimensionality reduction, misalnya principal component analysis (PCA). PCA membentuk representasi berdimensi lebih rendah dan tidak harus menghasilkan label cluster. Karena itu, clustering adalah contoh penting, bukan definisi yang mencakup seluruh unsupervised learning.

Tidak adanya label target juga tidak berarti tidak ada tujuan matematis. Algoritma tetap membutuhkan kriteria untuk menentukan struktur yang dianggap berguna.

**Rujukan:** Trask (2019), hlm. 13. Klarifikasi cakupan dan contoh PCA didukung oleh [Stanford Engineering Everywhere - CS229](https://see.stanford.edu/Course/CS229), bagian Course Description dan Lecture 14.

## 11. Parametric vs. nonparametric learning

Buku memperkenalkan perbedaan ini melalui cara pola disimpan di dalam model. Model parametric diibaratkan memiliki sejumlah tombol pengaturan yang tetap. Belajar dilakukan dengan menyesuaikan nilainya. Pada ilustrasi nonparametric, representasi dapat mengikuti informasi yang ditemukan dalam data.

**Pengertian yang perlu dipertahankan:**

- **Parametric:** untuk bentuk model yang telah ditetapkan, pola diwakili oleh sejumlah parameter yang tetap dan berhingga.
- **Nonparametric:** representasi atau keluwesan model tidak dibatasi oleh satu jumlah parameter tetap seperti itu; kebutuhan representasinya dapat bergantung pada data.

Buku menyederhanakan pendekatan pertama sebagai mencoba pengaturan dan pendekatan kedua sebagai menghitung. Analogi ini membantu pembaca membayangkan proses belajar, tetapi cara optimasi atau penggunaan hitungan tidak sendirian menentukan apakah suatu model parametric atau nonparametric.

**Klarifikasi dari referensi pendamping:** Catatan CS229 membandingkan linear regression dengan jumlah koefisien tetap dan locally weighted linear regression yang mempertahankan data pelatihan untuk membentuk prediksi lokal. Perbedaan pentingnya terletak pada representasi model, bukan pada apakah algoritma memiliki angka pengaturan.

**Rujukan:** Trask (2019), hlm. 14-15 dan 18; [Ng - CS229 Lecture Notes](https://cs229.stanford.edu/summer2023/cs229-notes1.pdf), bagian Locally weighted linear regression, hlm. 14.

## 12. Pendalaman: jumlah dan nilai parameter

**Jumlah parameter** dan **nilai parameter** adalah dua hal berbeda. Pada model parametric yang bentuknya telah ditetapkan, nilai parameter berubah saat belajar, tetapi jumlah parameter tidak harus bertambah ketika lebih banyak contoh pelatihan diberikan.

**Contoh tambahan:** Model sederhana $\hat{y}=a x+b$ memiliki dua parameter, yaitu $a$ dan $b$. Memberikan 10 atau 1.000 pasangan observasi tidak mengubah jumlah dua parameter tersebut selama bentuk modelnya tetap sama. Yang diperkirakan dari data adalah nilainya.

Pada neural network dengan arsitektur yang telah ditetapkan, bobot dan bias juga membentuk himpunan parameter berhingga. Jika peneliti mengubah arsitektur, jumlahnya dapat berubah. Perubahan desain ini perlu dibedakan dari penambahan contoh pelatihan pada arsitektur yang sama.

Istilah nonparametric **bukan berarti tanpa parameter**. Model masih dapat memiliki angka pengaturan dan representasi yang disimpan. Penyebutan "infinite" pada pengantar buku juga tidak berarti komputer menyimpan angka tak berhingga; pada kumpulan data berhingga, implementasinya tetap berhingga.

**Rujukan:** Trask (2019), hlm. 14-15 dan 18. Model $a x+b$ dan penjelasan arsitektur merupakan contoh pendamping.

## 13. Dua sumbu pengelompokan

Supervised/unsupervised dan parametric/nonparametric tidak merupakan satu deretan pilihan yang saling menggantikan. Keduanya menjelaskan aspek berbeda dan dapat dikombinasikan.

| | Parametric | Nonparametric |
| --- | --- | --- |
| Supervised | Belajar memprediksi target menggunakan representasi parameter yang tetap | Belajar memprediksi target dengan representasi yang dapat mengikuti data |
| Unsupervised | Belajar struktur data melalui parameter model yang tetap | Belajar struktur data melalui representasi yang dapat mengikuti data |

Matriks ini merangkum kerangka pengantar pada halaman 14. Nama kedua sumbu perlu dijawab terpisah: **apakah tersedia target untuk belajar**, dan **bagaimana pola direpresentasikan dalam model**?

Deep learning dalam cakupan buku merupakan kelompok model parametric. Neural network dapat digunakan pada tugas supervised maupun unsupervised. Penggunaan neural network karena itu tidak dengan sendirinya menentukan apakah tugasnya supervised.

**Rujukan:** Trask (2019), hlm. 14, 17-19.

## 14. Supervised parametric learning

Bagian ini menggunakan analogi mesin dengan tombol pengaturan. Informasi masukan diproses menurut posisi tombol, lalu menghasilkan prediksi. Saat hasil sebenarnya diketahui, prediksi dibandingkan dengan target dan nilai tombol disesuaikan.

Dalam model, tombol tersebut mewakili parameter. Parameter memengaruhi bagaimana masukan diterjemahkan menjadi keluaran. Nilai parameter pada suatu saat menyimpan pola yang telah dipelajari model.

Buku mencontohkan prediksi kemenangan tim baseball dengan beberapa informasi masukan. Sebagian informasi mungkin lebih relevan untuk prediksi dibandingkan informasi lain. Penyesuaian parameter diibaratkan sebagai perubahan sensitivitas terhadap informasi tersebut.

**Pendalaman:** Penyesuaian dalam implementasi dilakukan oleh prosedur belajar, bukan dengan manusia memutar tombol secara manual. Analogi sensitivitas juga tidak selalu berarti satu bobot dapat langsung ditafsirkan sebagai besarnya pengaruh kausal suatu faktor. Pada model yang lebih kompleks, keluaran muncul dari gabungan banyak operasi dan parameter.

**Rujukan:** Trask (2019), hlm. 15-16. Batas interpretasi sensitivitas merupakan pendalaman tambahan.

## 15. Predict, compare, learn

| Tahap | Apa yang dilakukan | Pertanyaan yang dijawab |
| --- | --- | --- |
| Predict | Mengolah masukan dengan parameter saat ini | Apa prediksi model? |
| Compare | Membandingkan prediksi dengan target yang diketahui | Seberapa berbeda hasilnya? |
| Learn | Menyesuaikan parameter berdasarkan masukan dan perbandingan | Bagaimana model perlu diperbaiki? |

Dalam ilustrasi buku, model memberikan prediksi peluang menang sebesar 98%, sedangkan pertandingan berakhir kalah. Perbandingan ini menjadi sinyal untuk menyesuaikan model agar prediksi berikutnya lebih sesuai dengan pola data.

Notasi pendamping untuk memisahkan dua tahap pertama adalah:

$$
\hat{y}=f_\theta(x),\qquad L=\ell(\hat{y},y)
$$

Fungsi $f_\theta$ menghasilkan prediksi, sedangkan $\ell$ mengukur ketidaksesuaian prediksi dan target. Tahap learn menggunakan informasi tersebut untuk mengubah $\theta$. Pilihan fungsi perbandingan bergantung pada tugas; persamaan ini tidak menetapkan algoritma pembaruan tertentu.

Alur ini penting karena **membuat prediksi saja belum berarti belajar**. Pada model yang digunakan tanpa pembaruan parameter, tahap prediksi dapat dijalankan tanpa tahap learn.

**Rujukan:** Trask (2019), hlm. 16. Notasi fungsi dan loss merupakan pendalaman tambahan.

## 17. Pendalaman: peluang prediksi dan hasil pertandingan

Ilustrasi 98% pada buku perlu dibaca dengan membedakan **peluang** dan **hasil yang teramati**.

## 18. Unsupervised parametric learning

Buku mengibaratkan pendekatan ini sebagai penggunaan beberapa tombol untuk setiap kelompok. Parameter diatur agar model mengenali pola masukan yang berkaitan dengan kelompok tertentu.

Ilustrasi buku menggunakan informasi pertandingan, seperti lokasi home/away dan jumlah penonton. Suatu observasi dapat memperoleh nilai keanggotaan untuk beberapa kelompok. Contoh yang ditampilkan adalah:

| Kelompok | Peluang keanggotaan dalam ilustrasi |
| --- | --- |
| Group 1 | 94% |
| Group 2 | 1% |
| Group 3 | 5% |

Ketiga angka tersebut berjumlah 100%. Pada ilustrasi itu, observasi memiliki keanggotaan paling kuat pada Group 1. Ini memperlihatkan gagasan **soft assignment**, yaitu menyatakan derajat keanggotaan, dibandingkan **hard assignment** yang menetapkan satu kelompok.

**Klarifikasi:** Nilai di antara 0 dan 1 tidak otomatis merupakan probabilitas yang sah. Interpretasinya bergantung pada definisi model; probabilitas untuk kelompok yang saling eksklusif dan mencakup seluruh kemungkinan harus berjumlah satu. Tidak semua algoritma unsupervised mengeluarkan probabilitas keanggotaan seperti contoh ini.

Angka pada tabel adalah ilustrasi dari buku, bukan hasil pelatihan model di notebook. Tidak tersedia label target kelas seperti pada supervised learning; struktur dipelajari melalui kriteria yang ditetapkan algoritma.

**Rujukan:** Trask (2019), hlm. 17. Istilah assignment dan syarat interpretasi probabilitas merupakan pendalaman tambahan.

## 19. Nonparametric learning

Buku menjelaskan pendekatan ini melalui metode yang menghitung pola dari data. Contohnya adalah pengamatan hubungan warna atau posisi lampu lalu lintas dengan kondisi kendaraan bergerak atau berhenti.

Proporsi empiris dapat ditulis dengan notasi pendamping:

$$
\hat{p}(\mathrm{go}\mid c)=\frac{n(\mathrm{go},c)}{n(c)}
$$

Di sini $n(c)$ adalah jumlah pengamatan kategori lampu $c$, dan $n(\mathrm{go},c)$ adalah jumlah pengamatan ketika kendaraan bergerak pada kategori tersebut. Penyebut harus lebih besar dari nol agar proporsinya dapat dihitung.

Buku mencontohkan proporsi 100% untuk lampu tengah dan 50% untuk lampu kanan pada pengamatan yang diilustrasikan. Nilai tersebut adalah ringkasan contoh buku, bukan estimasi yang dihitung dari dataset dalam notebook ini. Proporsi dari beberapa pengamatan juga tidak menjamin pola yang sama akan selalu terjadi pada pengamatan berikutnya.

Dalam analoginya, ukuran tabel hitungan dapat bertambah ketika informasi kategori yang disimpan bertambah. Buku menggunakan sifat tersebut untuk memperkenalkan gagasan representasi yang mengikuti data.

**Klarifikasi penting:** Menghitung frekuensi tidak otomatis membuat model nonparametric. Jika tiga kategori lampu telah ditetapkan dan masing-masing diwakili satu peluang bergerak, terdapat tiga parameter tetap. Berdasarkan pengertian jumlah parameter tetap, itu dapat menjadi model parametric. Contoh buku perlu dibaca sebagai analogi, bukan aturan umum untuk menggolongkan algoritma.

**Rujukan:** Trask (2019), hlm. 18. Klarifikasi terakhir merupakan penerapan definisi representasi parameter dari [CS229 Lecture Notes](https://cs229.stanford.edu/summer2023/cs229-notes1.pdf), hlm. 14.

| Training | Proses mempelajari atau menyesuaikan model dari data |
| Inference | Menggunakan model untuk menghasilkan prediksi atau keluaran |
| Cluster | Kelompok observasi berdasarkan pola yang ditentukan pendekatan pengelompokan |
| Generalization | Kemampuan menggunakan pola yang dipelajari pada contoh baru |

Buku menekankan bahwa istilah parameter cukup luas: dapat berupa bobot, hitungan, atau angka lain yang digunakan untuk mewakili suatu pola. Karena itu, **parameter bukan sinonim khusus untuk bobot neural network**.

Istilah loss, inference, dan generalization ditambahkan untuk memperjelas teori dan menyiapkan pembacaan bab berikutnya.
