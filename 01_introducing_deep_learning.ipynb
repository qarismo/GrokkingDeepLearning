## Ringkasan Chapter 1

Chapter 1 memperkenalkan alasan dan persiapan untuk mempelajari deep learning. Trask menjelaskan bahwa teknologi ini memungkinkan sebagian tugas yang memerlukan pengenalan pola dan pengolahan informasi diotomatisasi. Buku mencontohkan bidang penglihatan komputer, pengenalan suara, dan penerjemahan mesin untuk menunjukkan luasnya penerapan neural network. Potensi perubahan pekerjaan menjadi alasan lain untuk mempelajari bidang ini, di samping ketertarikan terhadap bagaimana mesin dapat menghasilkan perilaku yang tampak cerdas atau kreatif.

Pendekatan buku menekankan pemahaman mekanisme di balik deep learning. Pembaca memulai dari operasi sederhana dan membangun implementasi secara bertahap, sehingga mengetahui fungsi suatu alat sebelum menggunakannya melalui framework. Pengetahuan matematika tingkat lanjut tidak dijadikan syarat awal; konsep yang diperlukan dijelaskan dengan analogi dan contoh selama proses belajar. Pembelajaran berbasis proyek juga membantu menghubungkan teori dengan penggunaan praktis.

Bekal awal yang disebutkan adalah matematika dasar, pengetahuan Python, Jupyter Notebook, serta NumPy untuk operasi numerik. Pembaca juga dianjurkan memilih masalah yang menarik secara pribadi agar memiliki alasan konkret untuk terus belajar. Bab ini berakhir dengan kesiapan menuju konsep dasar pada Chapter 2. Chapter 1 sendiri belum memuat kode neural network atau eksperimen yang perlu direproduksi.

**Rujukan:** Trask (2019), Chapter 1, hlm. 3-8.

## 5. Why you should learn deep learning

### 5.1 Otomatisasi sebagian tugas yang memerlukan kecerdasan

Trask menggambarkan deep learning sebagai kelanjutan perkembangan alat yang membantu manusia memahami dan mengendalikan lingkungannya. Perbedaannya, banyak tugas yang dibantu berkaitan dengan pengolahan informasi, bukan hanya pekerjaan mekanis. Bidang yang disebut dalam bab ini meliputi computer vision, speech recognition, dan machine translation.

Neural network menjadi pendekatan yang dapat digunakan pada berbagai jenis masalah. Namun, contoh penerapan yang beragam tidak berarti satu model yang sama dapat langsung digunakan untuk seluruh tugas tanpa penyesuaian.

### 5.2 Menghadapi perubahan pekerjaan

Buku membahas kemungkinan otomatisasi pada sebagian pekerjaan terampil. Pandangan ini menjadi dorongan bagi pembaca untuk mengembangkan kemampuan baru dan memahami teknologi yang dapat memengaruhi cara kerja suatu profesi.

Pernyataan tentang perubahan pekerjaan perlu dibaca sebagai argumentasi dan prediksi penulis pada saat buku diterbitkan. Bab ini bukan laporan tentang tingkat penggantian pekerjaan pada tahun 2026. Dalam tugas ini, pelajaran yang diambil adalah pentingnya memahami teknologi dan menilai fungsi mana yang dapat dibantu oleh sistem otomatis.

### 5.3 Menjelajahi kreativitas

Alasan lain yang diberikan adalah rasa ingin tahu. Membangun sistem yang dapat meniru pola kreatif, misalnya gaya suatu lukisan, menjadi kesempatan untuk mempelajari hubungan antara data, pola, dan keluaran yang dihasilkan mesin.

**Pendalaman:** Keluaran yang terlihat kreatif tetap perlu dibedakan dari pengalaman, niat, atau kesadaran manusia. Kemiripan hasil tidak dengan sendirinya membuktikan bahwa proses mentalnya sama.

**Rujukan:** Trask (2019), hlm. 4. Penjelasan batas interpretasi merupakan pendalaman tambahan.

Dengan demikian, manfaat belajar deep learning bukan hanya menggunakan istilah "AI", melainkan mampu menjelaskan kemampuan yang ingin dibangun secara spesifik. Masalah yang jelas membantu menentukan apa yang perlu dipelajari dan apa yang perlu dibuktikan.

## 7. Will this be difficult to learn?

Trask mengurangi kesan bahwa pembaca harus menguasai seluruh teori sebelum dapat mulai membuat sesuatu. Buku dirancang dengan hasil belajar kecil yang muncul secara bertahap. Pembaca akan membuat neural network untuk prediksi pada Chapter 3 dan mempelajari perubahan yang membuat model belajar pada Chapter 4.

Strategi yang dijelaskan adalah mempelajari bagian kecil kode, mengingatnya, lalu menggunakan pengetahuan tersebut untuk memahami tahap berikutnya. Setiap tahap memberi konteks bagi tahap sebelumnya sehingga pembaca dapat melihat hubungan antara konsep dan implementasi.

**Pendalaman:** Untuk tugas reproduksi, mengingat kode sebaiknya disertai kemampuan menjelaskan perannya. Sebagai contoh, pada bab yang sudah memiliki kode, mahasiswa dapat menjawab: data apa yang masuk, operasi apa yang dilakukan, variabel apa yang berubah, dan apa arti hasilnya. Kemampuan menjelaskan ini lebih menunjukkan pemahaman daripada sekadar menyalin baris kode.

Pernyataan bahwa buku mudah dimulai bukan berarti seluruh materi tidak membutuhkan usaha. Bekal awal yang sederhana memungkinkan pembaca masuk ke pembahasan; pemahaman yang lebih dalam tetap dibangun melalui latihan dan pembacaan ulang.

## 9. Pendalaman teori: implementasi dasar dan framework

Implementasi dari dasar membantu memperlihatkan operasi yang sering disembunyikan oleh antarmuka framework. Sebaliknya, framework membantu mengelola pekerjaan berulang ketika implementasi menjadi lebih kompleks. Kedua pendekatan memiliki peran yang saling melengkapi dalam proses belajar.

| Aspek belajar | Implementasi dasar | Menggunakan framework |
| --- | --- | --- |
| Hal yang diamati | Operasi, variabel, dan aliran perhitungan | Komponen model dan cara menggunakannya |
| Manfaat untuk pemula | Memahami apa yang terjadi pada tiap langkah | Mengenal alat untuk menyusun model |
| Pertanyaan utama | Mengapa operasi ini dibutuhkan? | Bagaimana komponen ini digunakan dengan tepat? |
| Hubungan keduanya | Memberi dasar untuk memahami abstraksi | Menggunakan abstraksi untuk bekerja lebih efisien |

**Contoh tambahan:** Jika suatu fungsi memberikan keluaran numerik, pengguna yang memahami mekanismenya dapat menelusuri informasi masukan, operasi yang dilakukan, dan arti keluarannya. Tanpa pemahaman tersebut, memanggil fungsi mungkin berhasil secara teknis, tetapi penjelasan tentang hasilnya masih terbatas.

Untuk tugas ini, reproduksi kode pada bab berikutnya sebaiknya disertai penjelasan **masukan, proses, keluaran, dan alasan penggunaan operasi**. Chapter 1 memberikan dasar tentang mengapa kebiasaan belajar tersebut diperlukan, meskipun belum memberikan algoritmanya.

**Dasar pembahasan:** Trask (2019), hlm. 5-6. Tabel dan contoh merupakan sintesis pendamping.

| Jupyter Notebook | Menjalankan eksperimen secara bertahap dan memeriksa bagian perhitungan |

**Mengapa masalah pribadi?** Tujuan konkret membantu pembaca mempertahankan motivasi. Masalah tersebut menjadi alasan untuk mempelajari konsep baru dan mencoba prototipe. Memilih masalah adalah titik awal, bukan jaminan bahwa deep learning pasti menjadi metode yang paling sesuai.

## 13. Penyesuaian tugas ke Google Colab

Buku menggunakan Jupyter Notebook. Untuk tugas ini, lingkungan kerja yang dipilih adalah **Google Colab**. Dokumentasi resmi Google menjelaskan bahwa Colab merupakan layanan Jupyter Notebook yang dihosting dan dapat mengimpor notebook berformat `.ipynb`.

Notebook ini memakai **sel Text/Markdown** karena Chapter 1 tidak memuat listing kode. Ringkasan, tabel, penjelasan teori, dan referensi tetap dapat disusun di dalam notebook. Bab yang berisi kode nantinya dapat menggunakan sel kode beserta keluaran dan penjelasan hasilnya.

Cara membuka file ini: masuk ke Colab, pilih **File > Upload notebook**, lalu unggah file `.ipynb` ini. Setelah penyuntingan, simpan notebook dan unggah file hasilnya ke repository GitHub untuk tugas.

Tidak ada kebutuhan menjalankan eksperimen atau memilih GPU untuk isi Chapter 1 ini. Tidak ada dataset, instalasi library, atau output numerik yang harus dibuat agar bab tanpa kode terlihat memiliki reproduksi program.

**Rujukan penyesuaian platform:** [Google Colab - Pertanyaan Umum](https://research.google.com/colaboratory/intl/id/faq.html), bagian "Apa itu Colaboratory?" dan "Dapatkah saya mengimpor notebook Jupyter/IPython ke Colab?" Diakses 9 Oktober 2026.
